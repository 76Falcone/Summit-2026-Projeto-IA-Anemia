*Análise de Classificação utilizando Support Vector Machine (SVM)*

```python
import pandas as pd
import numpy as np
import os
from sklearn.preprocessing import LabelEncoder

# Buscando caminho correto
caminho_arquivo = ""
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        if filename.endswith('.csv'):
            caminho_arquivo = os.path.join(dirname, filename)
            print(f"Arquivo encontrado e carregado: {caminho_arquivo}")
            break

df = pd.read_csv(caminho_arquivo)
```

```python
print("--- Primeiras linhas do Dataset ---")
display(df.head())

print("\n--- Informações Gerais ---")
df.info()

print("\n--- Estatísticas Descritivas ---")
display(df.describe())
```

```python
print("--- Contagem de Dados Ausentes (Antes) ---")
print(df.isnull().sum())

for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())

df = df.drop_duplicates()

print("\n--- Contagem de Dados Ausentes (Depois) ---")
print(df.isnull().sum())
print("--- Classes Únicas na Coluna de Diagnóstico ---")
print(df['Diagnosis'].unique())

le = LabelEncoder()
df['Diagnosis_Encoded'] = le.fit_transform(df['Diagnosis'])

mapeamento = dict(zip(le.classes_, le.transform(le.classes_)))
print("\n--- Mapeamento do Encoding ---")
for chave, valor in mapeamento.items():
    print(f"{chave}: {valor}")

df = df.drop(columns=['Diagnosis'])

print("\n--- Dataset pronto para o SVM ---")
display(df.head())
```
